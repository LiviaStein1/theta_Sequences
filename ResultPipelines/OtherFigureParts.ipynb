```python
import numpy as np
import matplotlib.pyplot as plt

# 5 cycles, high resolution
cycles = 6
points_per_cycle = 1000
n_points = cycles * points_per_cycle

t = np.linspace(0, cycles, n_points)
theta_wave = np.sin(2 * np.pi * t)

fig, ax = plt.subplots(figsize=(19, 1))
fig.patch.set_alpha(0)
ax.set_facecolor('none')

ax.plot(t, theta_wave, color='#e2a9f1', lw=8)

ax.axis('off')
plt.tight_layout(pad=0)
plt.show()
fig.savefig("/Users/liviastein/STDP_Theta-1/Final_Figures/SineWaveFlatPink.svg", transparent=True, bbox_inches="tight", pad_inches=0)
```

```python
# Gaussian curve in specific colour
mean=0
std_dev=1.5
x = np.linspace(-5, 5, 1000)
gaussian_curve = (1/(std_dev * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mean) / std_dev) ** 2)

fig, ax = plt.subplots(figsize=(10, 2))
ax.plot(x, gaussian_curve, color='#5e17eb', lw=8)
ax.fill_between(x, gaussian_curve, color='#5e17eb', alpha=0.4)
ax.set_facecolor('none')
ax.axis('off')
plt.tight_layout(pad=0)
plt.show()
fig.savefig("/Users/liviastein/STDP_Theta-1/Final_Figures/GaussianCurveDviolet_fill.svg", transparent=True, bbox_inches="tight", pad_inches=0)
```

```python
from scipy.stats import skewnorm

# Parameters
mean = -1      # shifts the centre (location)
std_dev = 1.9   # controls spread (scale)
skew = 2.5        # 0 = symmetric (normal), + = right tail, - = left tail

x = np.linspace(-5, 5, 1000)
curve = skewnorm.pdf(x, a=skew, loc=mean, scale=std_dev)

fig, ax = plt.subplots(figsize=(10, 2))
ax.plot(x, curve, color='#e2a9f1', lw=8)
ax.fill_between(x, curve, color='#e2a9f1', alpha=0.4)
ax.set_facecolor('none')
ax.axis('off')
plt.tight_layout(pad=0)
plt.show()
```
